```python
df.rename(columns={"pcv":"packed_cell_volume","wc":"white_blood_cells_count","rc":"red_blood_cells_count","htn":"hypertension","dm":"diabetes_melinus","cad":"c_artery_disease","appet":"appettite","pe":"poda_odema","ane":"anemia"},inplace = True)
```

```python
df
```

```python
columns=['packed_cell_volume','white_blood_cells_count','red_blood_cells_count']
df[columns]= df[columns].apply(pd.to_numeric,errors='coerce')
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
for col in df.columns:
  if df[col].dtype=='object':
      df[col].fillna(df[col].mode()[0],inplace=True)
  else:
      df[col].fillna(df[col].mean(),inplace=True)
```

```python
df.isnull().sum()
```

```python
df['diabetes_melinus'].unique()
```

```python
df['c_artery_disease'].unique()
```

```python
df['classification'].unique()
```

```python
df['diabetes_melinus']=df['diabetes_melinus'].replace({' yes':'yes','\tno':'no','\tyes':'yes'})
```

```python
df['c_artery_disease']=df['c_artery_disease'].replace({'\tno':'no'})
```

```python
df['classification']=df['classification'].replace({'ckd\t':'ckd'})
```

```python
df['classification'].unique()
```

```python
df['classification'] = df['classification'].map({'ckd': 1, 'notckd': 0})
df['red_blood_cells'] = df['red_blood_cells'].map({'normal': 1, 'abnormal': 0})
df['plus_cells'] = df['plus_cells'].map({'normal': 1, 'abnormal': 0})
df['plus_cells_pumps'] = df['plus_cells_pumps'].map({'present': 1, 'notpresent': 0})
df['bacteria'] = df['bacteria'].map({'present': 1, 'notpresent': 0})
df['hypertension'] = df['hypertension'].map({'yes': 1, 'no': 0})
df['diabetes_melinus'] = df['diabetes_melinus'].map({'yes': 1, 'no': 0})
df['c_artery_disease'] = df['c_artery_disease'].map({'yes': 1, 'no': 0})
df['appettite'] = df['appettite'].map({'good': 1, 'poor': 0})
df['poda_odema'] = df['poda_odema'].map({'yes': 1, 'no': 0})
df['anemia'] = df['anemia'].map({'yes': 1, 'no': 0})
```

```python
from sklearn.model_selection import train_test_split
```

```python
x= df.drop('classification',axis=1)
y=df['classification']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
x_test.shape
x_train.shape
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dct = DecisionTreeClassifier()
```

```python
dct.fit(x_train,y_train)
```

```python
y_pred=dct.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report,f1_score,precision_score,recall_score
```

```python
print(confusion_matrix(y_test,y_pred))
```

```python
print(accuracy_score(y_test,y_pred))
```

```python
print(precision_score(y_test,y_pred))
```

```python
print(f1_score(y_test,y_pred))
```

```python
print(recall_score(y_test,y_pred))
```

```python
print(classification_report(y_test,y_pred))
```
